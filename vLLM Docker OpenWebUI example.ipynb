#### vLLM on GB10

vLLM is an open-source engine for serving large language models (LLMs) quickly and efficiently. You load a model into it, and it exposes an API (OpenAI-compatible) that your apps, scripts, or chat UIs can call. It's used for inference, meaning running the model, not for training it.

A typical vLLM server answers at an address like http://localhost:8000/v1, and you can check which models it's serving at /v1/models. 

vLLM and one model are already installed on your GB10 device. Use the code below to launch vLLM and serve the model. 

```python
# check the model's directory spelling with this:
ls ~/models/



# launch the vLLM container and serve the model
# copy and paste the docker command into the GB10 terminal 

# Update the following command with the exact spelling of the model:
# this example was stored as 
docker run -d \
  --gpus all \
  -p 8000:8000 \
  -v "$HOME/models/Qwen3-VL-8B-Instruct:/models/Qwen3-VL-8B-Instruct:ro" \
  -e HF_HUB_OFFLINE=1 \
  --ipc=host \
  --name qwen-vl-server1 \
  c1c9f6fd5c10 \
  /models/Qwen3-VL-8B-Instruct \
  --served-model-name Qwen/Qwen3-VL-8B-Instruct \
  --host 0.0.0.0 \
  --port 8000 \
  --gpu-memory-utilization 0.5 \
  --max-model-len 8192



# this example was downloaded as qwen3-VL-8b-instruct (some lower case letters)
docker run -d \
  --gpus all \
  -p 8000:8000 \
  -v "$HOME/models/qwen3-VL-8b-instruct:/models/qwen3-VL-8b-instruct:ro" \
  -e HF_HUB_OFFLINE=1 \
  --ipc=host \
  --name qwen-vl-server1 \
  c1c9f6fd5c10 \
  /models/qwen3-VL-8b-instruct \
  --served-model-name Qwen/Qwen3-VL-8B-Instruct \
  --host 0.0.0.0 \
  --port 8000 \
  --gpu-memory-utilization 0.5 \
  --max-model-len 8192


# open a new terminal window and run this to monitor the logs
docker logs -f qwen-vl-server1


# Troubleshooting- run this to start over
docker rm -fv qwen-vl-server1      # -v also deletes the stray anonymous volume
```

```python
# this will install WebUI to interact with vLLM

docker run -d -p 3000:8080 --add-host=host.docker.internal:host-gateway \
  -e OPENAI_API_BASE_URL=http://host.docker.internal:8000/v1 \
  -e OPENAI_API_KEY=EMPTY \
  -v open-webui:/app/backend/data --name open-webui \
  ghcr.io/open-webui/open-webui:main

# open another terminal window to check logs
docker logs -f open-webui
```
